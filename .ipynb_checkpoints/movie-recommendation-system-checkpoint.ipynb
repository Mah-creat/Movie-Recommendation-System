```python
#import libraries
import numpy as np
import pandas as pd
```

```python
#load data frames
movies = pd.read_csv('tmdb_5000_movies.csv')
credits = pd.read_csv('tmdb_5000_credits.csv')
movies.head()
```

```python
#returns a specified number of rows, string from the top
credits.head()
```

```python
#updates the content of two DataFrame by merging them together, using
#the specified method(s)
movies = movies.merge(credits, on = 'title')
#returns a specified number of rows, string from the top
movies.head(1)
#genres
#id
# keywords
#title
#overviw
#cast
#crewa
```

```python
#orgainsed all coloumn
movies = movies[['movie_id','title','overview','genres','keywords','cast','crew']]
#returns a specified number of rows, string from the top
movies.head()
```

```python
#Remove all rows wit NULL values from the DataFrame.It Returns a new
#DataFrame object unless the inplace parameter is set to True, in that
#case the dropna() method does the removing in the original DataFrame
#instead.
movies.dropna(inplace = True)
```

```python
# isnull() method to create a Boolean mask and then use the sum()
#method to count the number of True values.
movies.isnull().sum()
```

```python
#The duplicated() method returns a Series with True and False values
#that describe which rows in the DataFrame are duplicated and not. True
#can be counted with the sum() method. To count False (representing
#non-duplicate rows), negate with ~ and then use the sum() method.
movies.duplicated().sum()
```

```python
#An AST is a tree-like data structure that represents the syntactic
#structure of the code. It can be used for a variety of purposes, such
#as code analysis, code generation, and code transformation.The
#ast.Import() class represents an import statement in the AST
import ast
```

```python
import ast

def convert(text):
    L = []
    for i in ast.literal_eval(text):
        L.append(i['name']) 
    return L  # Moved outside the loop to return the WHOLE list
```

```python
# Apply the function to genres and keywords
movies['genres'] = movies['genres'].apply(convert)
movies['keywords'] = movies['keywords'].apply(convert)

# Check the results
movies.head()
```

```python
import ast

def convert3(text):
    L = []
    counter = 0
    for i in ast.literal_eval(text):
        if counter < 3:
            L.append(i['name'])
            counter += 1
        else:
            break  # Exit loop early once we have 3 items
    return L  # This must be aligned with the 'for' line
```

```python
movies['cast'] = movies['cast'].apply(convert)
movies.head()
```

```python
def fetch_director(text): # user define function which take text as an argument.
    L = []
    for i in ast.literal_eval(text):
        if i['job'] == 'Director':
            L.append(i['name'])
    return L
```

```python
# A quick way to remove spaces from all lists
movies['genres'] = movies['genres'].apply(lambda x: [i.replace(" ","") for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x: [i.replace(" ","") for i in x])
movies['cast'] = movies['cast'].apply(lambda x: [i.replace(" ","") for i in x])
movies['crew'] = movies['crew'].apply(lambda x: [i.replace(" ","") for i in x])
```

```python
# Converts the overview string into a list of words
movies['overview'] = movies['overview'].apply(lambda x: x.split())
```

```python
def collapse(L):
    if L is None:
        return []
    L1 = []
    for i in L:
        if i is not None:
            L1.append(i.replace(" ",""))
    return L1
```

```python
movies['cast'] = movies['cast'].apply(collapse)
movies['crew'] = movies['crew'].apply(collapse)
movies['genres'] = movies['genres'].apply(collapse)
movies['keywords'] = movies['keywords'].apply(collapse)
```

```python
# Combine the lists into the 'tags' column again
# Note: Ensure 'overview', 'genres', etc., have already been processed into lists
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
```

```python
# Create an independent copy to avoid the SettingWithCopyWarning
new_df = movies[['movie_id', 'title', 'tags']].copy()

# Now apply your transformations safely
new_df['tags'] = new_df['tags'].apply(lambda x: " ".join(x))
new_df['tags'] = new_df['tags'].apply(lambda x: x.lower())

# Check your final result
new_df.head()
```

```python
# The lambda function lambda x: x.lower() takes each element x 
# (which represents a string of tags) and converts it to lowercase 
# using the lower() method. This ensures uniformity in the casing 
# of the text, making it easier to compare and analyze.
new_df['tags'] = new_df['tags'].apply(lambda x: x.lower())

# Returns the first 5 rows to verify the transformation
new_df.head()
```

```python
import nltk
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()

def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)

# Apply stemming to the tags
new_df['tags'] = new_df['tags'].apply(stem)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000, stop_words='english')

# This creates a massive matrix of numbers (vectors)
vectors = cv.fit_transform(new_df['tags']).toarray()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vectors)
```

```python
# 2. Your existing stem function (now safe to use)
def stem(text):
    y = []
    if isinstance(text, list):
        input_data = text
    else:
        input_data = text.split()

    for i in input_data:
        y.append(ps.stem(i))

    return " ".join(y)
```

```python
# 3. Apply the function using .loc to ensure proper assignment
new_df.loc[:, 'tags'] = new_df['tags'].apply(stem)
```

```python
# View the result
new_df.head()
```

```python
# Import the CountVectorizer from Scikit-Learn
from sklearn.feature_extraction.text import CountVectorizer

# Instantiate the object with 5000 features and remove English stop words
# stop_words='english' removes common words like "the", "is", "and" 
cv = CountVectorizer(max_features=5000, stop_words='english')

# Transform the tags into a numerical matrix (vectors)
vectors = cv.fit_transform(new_df['tags']).toarray()

# Check the shape of the matrix (should be 4806 movies by 5000 words)
vectors.shape
```

```python
#cv.fit_transform(new_df['tags']): This part of the code applies the
#fit_transform method of the CountVectorizer object (cv) to the text
#data in the 'tags' column of the DataFrame new_df. This method first
#fits the CountVectorizer to the data (determining the vocabulary and
#other parameters) and then transforms the text data into a matrix of
#token counts.
#toarray(): The result of fit_transform is typically a sparse matrix
#representing the token counts. Calling .toarray() converts this sparse
#matrix into a dense NumPy array, where each row represents a document
#(or text entry) and each column represents a feature (or token).
vectors = cv.fit_transform(new_df['tags']).toarray()
```

```python
#vectors: This is the NumPy array containing the matrix of token
#counts obtained from the text data.
#shape: This is an attribute of NumPy arrays that returns a tuple
#representing the dimensions of the array.
vectors.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
#The cosine_similarity function from sklearn.metrics.pairwise computes
#the cosine similarity between pairs of samples in two given
#matricesThis line imports the cosine_similarity function from the
#sklearn.metrics.pairwise module.
#Cosine similarity is a measure of similarity between two non-zero
#vectors of an inner product space that measures the cosine of the
#angle between them.
#The cosine_similarity function takes the following parameters:X,Y, dense_output
```

```python
similarity = cosine_similarity(vectors)
```

```python
similarity.shape
```

```python
def recommend(movie):
    # Find the index of the movie title provided
    movie_index = new_df[new_df['title'] == movie].index[0]
    
    # Get the similarity scores for that specific index
    distances = similarity[movie_index]
    
    # Sort the scores while keeping track of original indices
    # We take [1:6] to skip the movie itself (which is at index 0)
    movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])[1:6]
    
    # Loop through the top 5 and print their titles
    for i in movies_list:
        # i[0] is the index of the recommended movie
        print(new_df.iloc[i[0]].title)

# --- TEST YOUR RECOMMENDER ---
recommend('Avatar')
```
